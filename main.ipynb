```python
import pandas as pd
import os
import sys

def get_formatted_results(base_path, abtype, window_length, abnormal_parameter, algorithms_to_display):
    """
    Reads a specific CSV file, processes the data for time=1000,
    and returns formatted strings for the metrics of interest.
    """
    # Construct the file path from the parameters
    dir_name = f"abtype{abtype}"
    file_name = f"abtype{abtype}_w{window_length}_t{abnormal_parameter}.csv"
    file_path = os.path.join(base_path, dir_name, file_name)

    # Initialize a dictionary to hold the results for each algorithm
    results = {}
    for alg in algorithms_to_display:
        results[alg] = {
            'CSS': '0.000 $\pm$ 0.000',
            'Sensitivity': '0.000 $\pm$ 0.000',
            'Specificity': '0.000 $\pm$ 0.000'
        }

    # Check if the file exists before trying to read it
    if not os.path.exists(file_path):
        print(f"Warning: File not found at {file_path}. Returning zeros.", file=sys.stderr)
        return results

    try:
        df = pd.read_csv(file_path)
    except Exception as e:
        print(f"Error reading {file_path}: {e}. Returning zeros.", file=sys.stderr)
        return results

    # Filter the DataFrame for the last captured time (1000)
    df_filtered = df[df['Captured Time'] == 1000].copy()

    # Process results for each specified algorithm
    for alg in algorithms_to_display:
        alg_data = df_filtered[df_filtered['Algorithm'] == alg]
        if not alg_data.empty:
            # Extract metrics and standard deviations.
            # Multiply by 100 to convert to percentage format.
            css = alg_data['Sum'].iloc[0] * 100
            css_std = alg_data['Sum (Std)'].iloc[0] * 100
            sens = alg_data['Sensitivities'].iloc[0] * 100
            sens_std = alg_data['Sensitivities (Std)'].iloc[0] * 100
            spec = alg_data['Specificities'].iloc[0] * 100
            spec_std = alg_data['Specificities (Std)'].iloc[0] * 100

            # Format the strings as "mean $\pm$ std"
            results[alg] = {
                'CSS': f"{css:.3f} $\pm$ {css_std:.3f}",
                'Sensitivity': f"{sens:.3f} $\pm$ {sens_std:.3f}",
                'Specificity': f"{spec:.3f} $\pm$ {spec_std:.3f}"
            }
    return results

def generate_latex_rows():
    """
    Main function to configure, fetch data, and print only the LaTeX table rows.
    """
    # =============================================================================
    # --- USER-CONFIGURABLE SECTION ---
    # =============================================================================

    # 1. Set the base path to your results directory.
    BASE_PATH = r'C:\Users\pault\Documents\3. AI and Machine Learning\2. Deep Learning\1c. App\Projects\CCPR_project\CSPR_2\results'

    # 2. List the algorithms you want to display in the table.
    ALGORITHMS_TO_DISPLAY = [
        'PA',
        'PA1',
        'PA2',
        'PA1_Csplit',
        'PA2_Csplit',
        'PA_L1', 
        'PA_L2',
        'PA1_L1',
        'PA1_L2',
        'PA2_L1', 
        'PA2_L2'
        # Add or remove other algorithm names from your CSVs here
    ]

    # 3. Define the structure and data sources for your table columns.
    TABLE_STRUCTURE = [
        {
            "header": "Fully Separable",
            "conditions": [
                {"name": "Up-trend, w=30, t=0.697", "params": {"abtype": 6, "window": 30, "abnormal_param": 0.697}},
            ]
        },
        {
            "header": "Partially Separable",
            "conditions": [
                {"name": "Up-trend, w=20, g=0.282", "params": {"abtype": 6, "window": 20, "abnormal_param": 0.282}},
            ]
        },
        {
            "header": "Least Separable",
            "conditions": [
                {"name": "Up-trend, w=15, g=0.005", "params": {"abtype": 6, "window": 15, "abnormal_param": 0.005}},
            ]
        },
    ]

    # =============================================================================
    # --- END OF USER-CONFIGURABLE SECTION ---
    # =============================================================================

    print(f"INFO: Reading data from base path: {BASE_PATH}\n", file=sys.stderr)

    # --- Data Fetching ---
    all_results = {}
    for block in TABLE_STRUCTURE:
        for condition in block['conditions']:
            params = condition['params']
            condition_name = condition['name']
            all_results[condition_name] = get_formatted_results(
                BASE_PATH,
                params['abtype'],
                params['window'],
                params['abnormal_param'],
                ALGORITHMS_TO_DISPLAY
            )

    # --- Generate and Print LaTeX Rows ---
    print("--- Copy the formatted rows below and paste them into your LaTeX table body ---")
    
    for alg in ALGORITHMS_TO_DISPLAY:
        # Sanitize algorithm name for LaTeX (e.g., replace underscore)
        display_alg_name = alg.replace('_', '\\_')
        
        row_cells = [display_alg_name]
        
        # Gather data across all configured conditions
        for block in TABLE_STRUCTURE:
            for condition in block['conditions']:
                res = all_results[condition['name']][alg]
                row_cells.append(res['CSS'])
                row_cells.append(res['Sensitivity'])
                row_cells.append(res['Specificity'])
        
        # Join all cell contents with the LaTeX separator '&' and end the line
        row_string = " & ".join(row_cells) + " \\\\"
        print(row_string)
        
    print("--- End of rows ---")

if __name__ == '__main__':
    generate_latex_rows()
```

```python
import os
from collections import Counter

def count_class_distribution(file_path):
    """
    Reads a wafer data file and counts the occurrences of each class label.

    The script assumes the first column of each row is the class label
    (e.g., '1' for normal, '-1' for abnormal).

    Args:
        file_path (str): The path to the input data file (e.g., 'wafer_TRAIN').
    """
    print(f"Analyzing file: {file_path}")

    # Check if the input file exists before proceeding.
    if not os.path.exists(file_path):
        print(f"--- Error: Input file not found at '{file_path}'\n")
        return

    # Use a Counter to automatically handle the counting of labels.
    class_counts = Counter()

    try:
        # Open the input file in read mode.
        with open(file_path, 'r') as f_in:
            # Process each line in the input file.
            for line in f_in:
                # Strip leading/trailing whitespace and skip empty lines.
                line = line.strip()
                if not line:
                    continue

                # The first element is the class label.
                # We split by comma and take the first part.
                label_str = line.split(',')[0]

                # Convert the label from a float string ('-1.0' or '1.0') to an integer.
                label = int(float(label_str))

                # The Counter will increment the count for the given label.
                class_counts[label] += 1

        # Print the results for the current file
        total_samples = sum(class_counts.values())
        print(f"--- Total Samples: {total_samples}")
        for label, count in sorted(class_counts.items()):
            class_name = "Normal" if label == 1 else "Abnormal"
            percentage = (count / total_samples) * 100 if total_samples > 0 else 0
            print(f"--- Class {label} ({class_name}): {count} samples ({percentage:.2f}%)")
        print("-" * 30 + "\n")

    except Exception as e:
        print(f"An error occurred while processing {file_path}: {e}\n")


# --- Main Execution ---
if __name__ == "__main__":
    # Define the input file paths.
    # This assumes the script is run from a directory that contains these files.
    # If not, you may need to provide the full path.
    train_file = 'wafer_TRAIN'
    test_file = 'wafer_TEST'

    print("--- Calculating Class Distribution for Wafer Datasets ---\n")
    count_class_distribution(train_file)
    count_class_distribution(test_file)
    print("Analysis complete.")
```
